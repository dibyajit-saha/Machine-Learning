```python
X = data[['Sorting Time']]
y =  data ['Delivery Time']
```

```python
model = LinearRegression()
model.fit(X,y)
```

```python
m = model.coef_
c = model.intercept_

print ("Slope = ", m)
print ("Constant = ", c)
```

```python
# Prediction
X1 = 5

y = [(m*X1) + c]
y
```
